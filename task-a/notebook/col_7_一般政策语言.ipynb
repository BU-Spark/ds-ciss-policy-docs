```python
def printColor(doc, idxs):
    idxs = sorted(idxs, key=lambda x: x[0], reverse=True)
    for idx in idxs:
        doc = doc[:idx[0]] + doc[idx[0]:idx[1]] + doc[idx[1]:]
        doc = f'{doc[:idx[0]]}\033[30;103m{doc[idx[0]:idx[1]+1]}\033[0m{doc[idx[1]+1:]}'
    print(doc)
```

```python
import os
dirSample = os.path.join(os.getcwd(), '..', 'sample')
if not os.path.exists(dirSample):
    print('No Sample Found')
    quit()
files = os.listdir(dirSample)
files = [f for f in files if f.endswith('.txt')]
with open(os.path.join(dirSample, files[0]), 'r') as f:
    doc = f.read()
    printColor(doc, [(218, 432)])
```
