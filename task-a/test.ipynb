```python
docs = """
【法宝引证码】 CLI.12.1092849
原文链接：https://www.pkulaw.com/lar/448d1430be882ed0d23a26d5b83256dcbdfb.html
贵港市人民政府办公室关于印发贵港市公共资源交易管理暂行办法的通知
贵港市人民政府办公室关于印发贵港市公共资源交易管理暂行办法的通知

（贵政办发〔2015〕44号）

各县、市、区人民政府，市人民政府各组成部门、各直属机构：

　　《贵港市公共资源交易管理暂行办法》已经市四届人民政府第35次常务会议审议通过，现印发给你们，请认真贯彻执行。
贵港市人民政府办公室

2015年7月3日
　　贵港市公共资源交易管理暂行办法
　第一章　总　则
　　第一条　为完善公共资源市场配置机制，提高公共资源市场配置效益，规范公共资源交易监督管理，根据《中华人民共和国招标投标法》、《中华人民共和国政府采购法》、《中华人民共和国拍卖法》、《中华人民共和国土地管理法》、《中华人民共和国招标投标法实施条例》、国土资源部《探矿权采矿权招标拍卖挂牌管理办法（试行）》和《招标拍卖挂牌出让国有建设用地使用权规定》（国土资源部第39号令）等相关法律、法规、规章，按照《广西壮族自治区关于进一步加强我区公共资源招标投标监督管理工作的意见》（桂政发〔2011〕75号）等文件精神，结合我市实际，制定本办法。
　　第二条　本办法适用于市本级按规定必须进入贵港市公共资源交易服务中心（以下简称“市交易服务中心”）进行交易的政府投（融）资或使用国有资金的工程建设项目（房屋建筑和市政基础设施、交通、水利等）、政府采购、国有土地使用权和矿业权出让、国有产权交易、医疗器械及药品采购、公共资源（资产）经营权和公共产品（服务）经营权出让等公共资源交易活动及其监督管理活动。

　　本办法所称公共资源，是指国家机关、事业单位和被授权的组织所有或者管理的，具有公有性、公益性的资源。
　　第三条　公共资源交易应当遵循公开、公平、公正和诚实信用以及竞争择优的原则。招标人、采购人、投标人、评审专家、代理机构（中介机构）等公共资源交易主体要依法参与公共资源交易活动，并接受贵港市公共资源交易监督管理办公室（以下简称“市公管办”）和行业行政主管部门的监督管理。
　　第四条　公共资源交易实行目录管理。列入《公共资源集中交易监管目录》（以下简称《目录》）的交易项目应当进入集中交易统一平台，采用招标、拍卖、竞价、挂牌、竞争性谈判、询价、单一来源采购等方式进行交易。具体交易方式由行政审批主管部门依据法律、法规和有关规定确定。

　第二章　组织机构及职责
　　第五条　贵港市公共资源交易监管工作领导小组（以下简称“领导小组”）作为全市公共资源交易管理的议事协调机构，对全市公共资源交易工作中重大事项进行研究、指导和决策，负责审议全市公共资源交易监管相关的综合规范性文件、管理制度；协调解决全市公共资源交易监管工作的重大问题。

　　发改、财政、住建、交通、水利、国土资源、国资、卫计委等行业行政主管部门按职能职责分工，依法履行行业内公共资源交易项目的监督管理，对公共资源交易活动主要履行以下职责：

　　（一）贯彻执行与公共资源交易监督管理有关的法律、法规和自治区有关规定；

　　（二）根据行业监督管理职能，依法对公共资源交易活动履行行政审批审核职责，指导、监督本行业公共资源交易活动的实施；

　　（三）负责制订涉及本行业公共资源交易项目目录及相关交易程序，并及时更新；

　　（四）负责对本行业交易公告、交易条件、交易文件等资料进行合规合法性审查或备案；

　　（五）负责对评标（审）专家抽取活动和开标、评标、竞价、拍卖等进场交易过程进行现场监督；

　　（六）负责对交易情况报告书和合同备案进行监督，对合同履约情况进行后续监管；

　　（七）负责受理公共资源交易中的违法违规行为和涉及公共资源交易的投诉、举报，办理其他监管部门转交的投诉举报，并依法进行调查，会同有关部门查处违反法律、法规及规章的行为；

　　（八）建立本行业的评标（审）专家库，健全对评标（审）专家的培训、考核、评价和信誉管理、档案管理制度，并指导、监督评标（审）专家库的使用；

　　（九）负责本行业各类中介机构及其从业人员、投标人（供应商、竞买人）等市场主体诚信体系建设；

　　（十）法律、法规及规章应当履行的其他职责。

　　监察机关依法对与公共资源交易活动有关的监察对象实施监察，并履行以下职责：

　　（一）对相关监督管理部门及其工作人员履行职责情况和执行本办法情况实行监督，受理涉及监察对象的投诉举报；

　　（二）对相关监督管理部门受理和处理涉及公共资源交易活动的举报投诉情况实行监督；

　　（三）负责查处相关监督管理部门及其工作人员在公共资源交易活动中的违纪违法行为；

　　（四）法律、法规及规章规定应当履行的其他职责。

　　市法制办依法受理涉及公共资源交易的行政复议和审查复核交易文件并提出处理意见等事项。

　　市审计局依法对各类公共资源交易项目实施审计监督。
　　第六条　贵港市公共资源交易监督管理办公室（以下简称“市公管办”）负责对市本级及三区进入市公共资源交易平台的公共资源交易活动实施综合指导、监督管理和统一协调工作，对公共资源交易活动主要履行以下职责：

　　（一）负责贵港市公共资源交易监管工作领导小组办公室的日常工作；

　　（二）贯彻执行公共资源交易活动的有关国家法律、法规和规章，依据法律、法规和本市实际，会同相关行业行政主管部门制定市公共资源交易工作规章制度和管理规定等；

　　（三）负责对行业建立的专家库进行备案，会同有关部门建立全市招标（政府采购）代理机构信用档案库、工程预算编制和审核中介机构信用档案库、公共资源交易市场竞争主体（投标人、供应商、竞买人等）诚信库；

　　（四）对列入《目录》的公共资源交易全过程实行监督管理，督促和指导市交易服务中心为全市公共资源集中交易集中监督管理提供相关服务；

　　（五）协调行业行政主管部门与市交易服务中心有关工作，及时转办、交办、督办公共资源交易各方的异议、质疑、投诉和举报；

　　（六）对行业行政主管部门履行监管职责进行监督；

　　（七）承担市政府交办的其他工作事项。
　　第七条　贵港市公共资源交易管理中心负责市本级及三区公共资源交易活动的管理工作。
　　第八条　市政务办负责组织工程建设项目、政府采购、国有土地使用权和矿业权出让、医疗器械及药品采购和产权交易等进入市交易服务中心交易，并确保交易过程依法依规进行。

　　贵港市公共资源交易服务中心是市政府指定的市级公共资源交易场所和服务平台，主要履行以下职责：

　　（一）负责制定公共资源交易现场管理规章制度、工作流程并组织实施，为公共资源交易活动提供交易平台和相关服务；

　　（二）负责收集、贮存和发布有关公共资源交易政策法规、规章制度以及进场交易项目、交易当事人、招标（政府采购）代理机构等信息；

　　（三）负责公共资源交易平台建设及日常管理维护，负责建立和完善公共资源交易电子招投标系统，对公共资源交易全程电子监控系统进行维护管理，建设和管理公共资源交易网站；

　　（四）负责按照有关规定受理、发布各类招标、政府采购、拍卖、出让等交易信息，查验进场交易项目的相关手续和参与交易活动各方的主体资格；

　　（五）受行业行政主管部门委托对评标（审）专家库进行维护，提供评标（审）专家抽取服务，对评标（审）专家的出勤情况和评标活动进行记录和见证，并将记录抄送市公管办；

　　（六）负责保障交易活动场所的秩序和安全，为实施监督管理提供条件，发现交易活动中违法、违规行为，及时报告行业行政主管部门并协助调查；

　　（七）负责对交易（招标、政府采购、拍卖、挂牌出让）文件等档案材料及进场交易活动的各类文字、音像、图片等资料的收集、整理、立卷、归档管理，按规定为相关部门提供档案查询服务；

　　（八）负责公共资源交易信息统计、分析及上报工作，并及时反馈市公管办；

　　（九）负责统一代售交易文件、代收代退投标保证金；

　　（十）承担市政府交办的其他工作事项。

　第三章　交易目录及交易程序
　　第九条　下列项目应当列入公共资源交易《目录》，实行集中交易和集中监管：

　　（一）国有（集体）企业的产权（股权）、知识产权；

　　（二）有限责任公司和非上市股份有限公司中的国有（集体）物权、债权、知识产权和股权；

　　（三）行政机关、社会团体、事业单位的国有物权、债权、知识产权和资产的租赁权；

　　（四）政府投资工程建设项目和政府采购项目以及市属国有企业使用政府投资或补助资金建设的项目；

　　（五）国家（集体）所有的自然资源使用权；

　　1.国有建设用地使用权；

　　2.水资源、荒地等自然资源使用权；

　　3.矿业权。

　　（六）政府特许经营权；

　　1.市政公用设施使用权；

　　2.出租汽车经营权；

　　3.主要污染物排放权等各类环境资源交易项目；

　　4.公共空间户外广告经营权；

　　5.城市道路和公共场地设置的停车泊位经营权；

　　6.公共设施（包括道路、桥梁等）的冠名权；

　　7.城市供气、供水及污水、废弃物处理等特可经营权。

　　（七）司法机关和行政执法部门的罚没资产处置项目；

　　（八）法院执行资产的拍卖项目；

　　（九）涉及国有（集团）权属有偿使用的其他交易品种和相关公共资源衍生权；

　　1.公办学校后勤社会化承包权；

　　2.市政设施维修承包权；

　　3.道路清扫作业承包权；

　　4.城市道路养护承包权；

　　5.政府投资项目中介服务机构的选择。

　　（十）国家、自治区和市人民政府规定应当列入《目录》的其他交易项目。
　　第十条　市本级公共资源交易《目录》的制定、修订由各行业行政主管部门根据上级要求和实际需要拟定，报经市人民政府批准后公布。三区公共资源交易《目录》的制定、修订由三区人民政府报市政府批准后公布。
　　第十一条　列入《目录》的公共资源交易项目应当进入市交易服务中心进行集中交易，接受集中监管。按规定应当进场交易而未进场交易或规避进场交易的，有关部门不得为其办理行政许可、资金拨付、产权过户和使用等手续，同时监察部门应对该项目的责任单位及责任人进行责任追究，构成犯罪的，依法移送司法机关处理。
　　第十二条　列入《目录》的下列公共资源交易项目，应当按照法律法规和政府的有关规定，确定代理机构代理交易事宜，招标人具有编制招标文件和组织开标能力的，可以自行办理招标事宜。

　　（一）投资限额以上的依法应当进行公开招标、邀请招标的建设工程项目；

　　（二）纳入集中采购目录的政府采购项目以及市属非竞争性国有独资企业通用类货物采购项目；

　　（三）国有产权及股权交易项目；

　　（四）政府确定的其他公共资源交易项目。

　　前款所称投资限额的规定，由政府根据相关法律法规确定。
　　第十三条　所有进场交易项目的进场交易受理、信息发布、报名受理、资格审查、评标（审）委员会组建、开标、评标、挂牌、拍卖、竞价等交易活动必须在市交易服务中心进行。
　　第十四条　进场公共资源交易项目一般按下列程序进行：

　　（一）申请登记。对符合本办法规定范围的交易项目，公共资源交易项目单位（包括招标人、采购人、竞卖人、出让人和出资人或者代表，以下简称“交易项目单位”）向行业行政主管部门提出项目交易申请。

　　行业行政主管部门应当按照法定程序和时限要求进行审核，对不符合法律、法规规定条件和要求的项目，应当拒绝受理交易申请，并说明理由。

　　行业行政主管部门审核完成后，交易项目单位向市交易服务中心提出进场交易申请，并提交项目批复及相关资料，市交易服务中心审验交易项目进场资格，对具备进场条件的交易项目进行备案登记后，向市公管办报备。

　　（二）信息发布。交易项目单位将已经审核批准的交易（招标、政府采购、拍卖、挂牌出让等）公告以及澄清、说明、补正等交易信息按有关规定在指定媒体发布，并在市公共资源交易服务中心门户网站和电子显示屏上同步发布。

　　（三）交易保证金。投标人（供应商、竞买人）根据交易（招标、政府采购、拍卖、挂牌出让等）文件规定要求，向市公共资源交易服务中心交纳交易保证金，市公共资源交易服务中心设立保证金专户，按相关规定进行管理，专人负责交易保证金的收取、退还操作。

　　（四）场地预约。交易项目单位根据交易项目的类型、规模提前向市交易服务中心提出场地预约申请，由市交易服务中心统一安排。

　　（五）评标（审）专家抽取。评标（审）专家应在有关行业行政主管部门、交易项目单位的现场监督下，由市交易服务中心组织在评标（审）专家库中随机抽取。

　　（六）开标、挂牌、拍卖。按交易（招标、政府采购、拍卖、挂牌出让等）文件规定的时间在市交易服务中心进行，由交易项目单位或委托的代理机构组织实施，投标人（供应商、竞买人）持有效证明材料参加，行业行政主管部门派人现场全程监督。

　　（七）评标（审）。交易项目单位依法组建项目评标（审）委员会进行评标（审），行业行政主管部门派员现场监督。评标（审）区实行封闭管理，无关人员不得随意进入评标（审）区。

　　（八）结果公示或公告。交易结果应按有关规定在指定媒体公示或公告，并在市公共资源交易服务中心门户网站和电子显示屏上同步发布，接受社会监督。

　　（九）成交确认。交易结果公示或公告期满无异议，由招标人（采购人、出让人）出具中标通知书（成交确认书），报行业行政主管部门备案。

　　（十）合同签订、备案。中标通知书（成交确认书）签发后， 招标人（采购人、出让人）和中标人（买受人）应在规定的时间内签订书面合同，报行业行政主管部门备案，同时向市公管办报备。

　　具体的公共资源交易程序和操作规则，按照交易类别和内容，由各相关行政主管部门根据现行法律、法规、规章另行制订，报市公共资源交易监管工作领导小组批准后公布执行。

　第四章　交易管理
　　第十五条　公共资源交易应坚持集中交易、集中监管原则，实行统一进场交易、统一信息发布、统一交易规则、统一监督管理。
　　第十六条　公共资源交易信息应按相关法律法规和行业的规定在指定的媒介发布，并同时在公共资源交易网络信息平台发布。

　　交易中心应建立网络信息平台，按规定及时发布公共资源交易信息。
　　第十七条　进场交易的公共资源交易项目，应当符合法律、法规规定的交易条件。有下列情形之一的，不得受理：

　　（一）未经依法审批、核准的；

　　（二）权属有争议的；

　　（三）被依法采取限制措施的；

　　（四）法律、法规规定禁止交易的其他情形。
　　第十八条　交易项目单位应把资格预审结果和招标文件以及招标文件的澄清和修改报送同级行业行政主管部门审查或备案。

　　交易项目单位可以根据项目需要，在招标文件中对投标人提出资格条件，但提出的资格条件应当有利于促进公平竞争，符合项目的合理需求，不得以不合理的条件排斥、歧视潜在投标人。
　　第十九条　公共资源交易资格审查委员会、评标委员会、谈判小组、询价小组的专家成员应当在评审专家库里随机抽取。
　　第二十条　本办法第十二条规定的建设工程项目和政府采购项目，按招标文件确定的评标办法和中标原则实行有效最低价或综合评分最高得分者竞得；国有产权及股权交易项目，实行有效最高价竞得。
　　第二十一条　所有进入交易中心交易的项目应当实行投标担保、低价风险担保和履约担保。
　　第二十二条　有下列情形之一的，应当中止交易：

　　（一）因交易系统故障致使交易不能进行的，或者因不可抗力致使交易不能进行的，应当中止公共资源交易。

　　（二）交易期间的公共资源权属存在异议，尚未依法确定权属的，应当中止公共资源交易。
　　第二十三条　有下列情形之一的，应当终止交易：

　　（一）人民法院、仲裁机构或者有关行政执法机关确认交易项目单位对其委托交易的项目无处理权的；

　　（二）依法应当终止交易的其他情形。
　　第二十四条　交易项目单位应当遵守本办法规定，应进入交易中心平台交易的项目全部进入，并及时确认交易信息，签订交易合同，履行合同备案、交易项目验收等职责。交易项目单位不得有下列行为：

　　（一）采取化整为零或者以其他任何方式进行场外交易；

　　（二）擅自中止、终止交易；

　　（三）单方面拒绝签订合同或者提出额外附加条件；

　　（四）阻挠和限制投标人自由进入交易服务中心平台；

　　（五）与交易竞争主体或者评审委员会成员恶意串通。
　　第二十五条　公共资源交易竞争主体（以下简称竞争主体）应当遵守交易活动的程序与规定，不得有下列行为：

　　（一）以他人名义交易或者以其他方式弄虚作假骗取项目竞得；

　　（二）恶意串通或者通过行贿等违法手段谋取竞得；

　　（三）捏造事实、伪造材料，或者以非法手段获取证明材料进行质疑或者投诉的；

　　（四）擅自放弃项目竞得资格、不签订交易合同、不提交履约保证金和低价风险保证金等担保，或在项目实施中降低技术标准。
　　第二十六条　招标（政府采购）代理机构等中介机构应当遵守交易活动的程序与规定，不得有下列行为：

　　（一）泄露应当保密的与交易活动有关的情况和资料；

　　（二）与交易项目单位、交易竞争主体串通损害国家利益、社会公共利益或者他人合法权益；

　　（三）在交易过程中接受贿赂或者获取其他不正当利益；

　　（四）在有关部门依法实施的监督检查中提供虚假情况；

　　（五）隐匿、销毁应当保存的交易文件或者伪造、编造招标文件。
　　第二十七条　评标（审）专家成员应当按照招标文件规定的标准和方法进行评审，不得有下列行为：

　　（一）私自接触交易竞争主体和收受财物或其他利益；

　　（二）向交易项目单位征询确定交易竞得者的意向；

　　（三）接受任何单位或者个人明示或者暗示提出的倾向；

　　（四）排斥特定竞争主体的要求；

　　（五）擅离职守等其他渎职行为。
　　第二十八条　交易项目单位和公共资源交易竞得者，应当按照交易文件的要求及时签订合同，不得另行签订背离合同实质性内容的其他协议。
　　第二十九条　交易项目单位应当在公共资源交易合同签订之日起七个工作日内，向行业行政主管部门和市公管办提交合同复印件和书面总结。

　第五章　监督管理
　　第三十条　对公共资源交易主要环节实行行业行政主管部门、行政监察部门、市公管办共同监管。采取场内监督、专项监督和社会监督相结合的方式，并通过现场和视频、音频监控结合的监督手段，对交易全程实行有效监督，形成互为制约、监督有力的监管机制。
　　第三十一条　市公管办会同有关行业行政主管部门建立完善中介机构市场准入和退出机制，加强对招标（政府采购）代理机构、预算编制（评审协作）机构等中介机构的规范管理和信用考核，促使中介机构依法依规开展经营活动，公平竞争。
　　第三十二条　市公管办会同有关行业行政主管部门整合专家资源，接受自治区招标投标综合评审专家库的管理，实现专家库资源共享。各有关行业行政主管部门严格评标专家资格认定，建立健全对评标（审）专家的培训、考核、评价和信誉管理、档案管理制度，根据实际需要和考核情况及时对评标（审）专家进行调整和补充，实行动态管理。
　　第三十三条　市公管办会同有关行业行政主管部门研究制订各类市场主体不良行为标准、信用信息管理办法和信用评价办法，建立健全信用信息和诚信体系信息的收集发布制度。有关行业行政主管部门依法查处投诉举报和各类违法、违规案件，应及时将处理意见或结果抄送市公管办，并按照管理权限和属地管理原则，负责对公共资源交易各方主体的诚信行为进行检查，收集、审核、记录从业单位和人员的信用信息，建立市场各方主体信用档案，向市公管办报备。
　　第三十四条　市公管办和行业行政主管部门要加强对相关单位贯彻国家法律法规和执行《目录》情况的监督检查，加强对项目单位是否擅自进行场外交易，是否按规定程序和规则组织交易活动进行监督检查。
　　第三十五条　列入《目录》的公共资源交易项目必须进入交易服务中心平台进行交易，接受集中监管。暂不进入交易服务中心平台进行交易的项目，市公管办与行业行政主管部门商定监督管理事宜。
　　第三十六条　行政主管部门的监察机构要加强对行业行政主管部门履行监督管理职责情况的监察；对交易项目单位执行《目录》情况和组织交易活动过程的监察；对交易服务中心的交易服务活动情况的监察；对行政机关和国家工作人员在交易活动中的行为进行执法监察。
　　第三十七条　参与公共资源交易活动的交易项目单位、投标人（供应商、竞买人）、招标（政府采购）代理机构、评标（审）专家及其他相关组织的人员违反公共资源交易活动（招标、政府采购、拍卖、挂牌出让等）的有关国家法律法规和行政规章规定的，由相关的行业行政主管部门、纪检监察机关按照有关规定追究其责任；构成犯罪的，移送司法机关依法追究其法律责任。

　第六章　附　则
　　第三十八条　涉及国家安全、国家秘密、抢险救灾等不适宜纳入集中交易、集中监督管理的交易项目，应经同级人民政府、政府主管部门审批、核准后执行。
　　第三十九条　法律、行政法规、部门规章对公共资源活动有其他规定的，从其规定。之前本市发布的有关规范性文件，与本办法不一致的，按照本办法执行。
　　第四十条　县（市）区公共资源交易管理工作参照本办法执行。
　　第四十一条　本办法由贵港市公共资源交易监督管理办公室负责解释。
　　第四十二条　本办法自发布之日起施行，试行期一年。
"""
```
