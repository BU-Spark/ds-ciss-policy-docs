```python
import os
cwd = os.getcwd()
print(cwd)
```

```python
def getTitle(line):
    if len(line) > 0:
        return True, line
    else:
        return False, ''
```

```python
docs = next(os.walk(os.path.join(cwd, 'sample')))[2]
titles = []
fails = []
for doc in docs:
    if not doc.endswith('.txt'):
        print('skip', doc)
        continue
    with open(os.path.join(cwd, 'sample', doc), 'r') as f:
        content = f.read()
        lines = content.split('\n')
        linePrev = ''
        for line in lines:
            if linePrev.startswith('原文链接'):
                success, title = getTitle(line)
                if success:
                    titles.append(title)
                else:
                    fails.append(line)
            linePrev = line
print(len(titles), '/', len(docs))
print(titles)
print(fails)
```
