```python
import os
docs = os.listdir(os.path.join(os.getcwd(), 'sample'))
print(f'Found {len(docs)} Samples')
errorCount = 0
for doc in docs:
    if not doc.endswith('.txt'):
        print(f'Invalid File: {doc}')
        errorCount += 1
print(f'Found {errorCount} Invalid Samples')
```

```python
import os
errorCount = 0
dataDir = os.path.join(os.getcwd(), '..', 'data')
regions = next(os.walk(dataDir))[1]
for region in regions:
    years = next(os.walk(os.path.join(dataDir, region)))[1]
    for year in years:
        docs = next(os.walk(os.path.join(dataDir, region, year)))[2]
        for doc in docs:
            if not doc.endswith('.txt'):
                print(f'Invalid File: {region}/{year}/ {doc}')
                errorCount += 1
print(f'Found {errorCount} Invalid Samples')
```
