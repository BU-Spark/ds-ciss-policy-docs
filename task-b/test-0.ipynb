```python
import os
cwd = os.getcwd()
print(cwd)
```

```python
mapL1 = {}
mapL2 = {}
trimLabel = lambda x: x.strip().split(' ')[0]
with open(os.path.join(cwd, 'mapping.csv')) as f:
    for line in f:
        line = line.strip().split(',')
        assert len(line) == 5
        labelL1, labelL2 = '', ''
        if len(line[3]) > 0:
            labelL1 = line[3]
        if len(line[4]) > 0:
            labelL2 = line[4]
        if len(line[0]) > 0:
            mapL1[trimLabel(line[0])] = labelL1
            mapL2[trimLabel(line[0])] = labelL2
        if len(line[1]) > 0:
            mapL1[trimLabel(line[1])] = labelL1
            mapL2[trimLabel(line[1])] = labelL2
        if len(line[2]) > 0:
            mapL1[trimLabel(line[2])] = labelL1
            mapL2[trimLabel(line[2])] = labelL2
len(mapL1), len(mapL2)
```

```python
import pandas as pd
df = pd.read_csv(os.path.join(cwd, 'policies_all.csv'))
df['type'] = df['type'].astype(str)
df.head()
```

```python
# Documents with more than one label (splitted with a comma)
df[df['type'].apply(lambda x: len(x.split(','))) > 1]
```

```python
# use the first label 
df['type'] = df['type'].apply(lambda x: x.split(',')[0])
```

```python
df['label_L1'] = df['type'].map(mapL1)
df['label_L2'] = df['type'].map(mapL2)
df
```

```python
print("All records: {}. Record with no type (or 404): {}. Records with no L1 label: {}. Records with no L2 label: {}".format(len(df), len(df[(df['type'].isnull())|(df['type']=='404-NotFound')]), len(df[df['label_L1'].isnull()]), len(df[df['label_L2'].isnull()])))
```

```python
df_noMap = df[~((df['type'] == 'nan') | (df['type']=='404-NotFound')) & df['label_L1'].isnull()].copy()
df_noMap.to_csv(os.path.join(cwd, 'policies_noMap.csv'), index=False)
df_noMap.shape
```

```python
df_noMap
```

```python
def matchLongest(x):
    res = ''
    x = x.strip()
    for key in mapL1:
        if x.startswith(key) and x[len(key):] in mapL1 and len(key) > len(res):
            res = key
    return res if res != '' else x
df_noMap['type'] = df_noMap['type'].apply(matchLongest)
df_noMap['label_L1'] = df_noMap['type'].map(mapL1)
df_noMap['label_L2'] = df_noMap['type'].map(mapL2)
```

```python
df.update(df_noMap)
df_noMap = df[~((df['type'] == 'nan') | (df['type']=='404-NotFound')) & df['label_L1'].isnull()].copy()
df.to_csv(os.path.join(cwd, 'policies_label.csv'), index=False)
df_noMap.to_csv(os.path.join(cwd, 'policies_noMap.csv'), index=False)
df_noMap.shape
```
